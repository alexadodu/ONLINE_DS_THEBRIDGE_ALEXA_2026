## Ejercicio 10
Convierte el programa del ejercicio 2 en una función al que se le pase el valor a buscar como un argumento posicional

`````python
ejer_3 = ["Un", "árbol", "binario", "es", "una", "estructura", "de", "un", "tipo", "particular", "a", "veces", "no", "es", "ni", "binario"]
duplicados_ejer = ["un", "es", "binario"]

duplicado_dict = {}

def busqueda_duplicados(duplicado):
    for indice, palabra in enumerate(ejer_3):
        if palabra == duplicado:
            if palabra in duplicado_dict.keys():
                duplicado_dict[palabra].append(indice)
            else:
                duplicado_dict[palabra]= [indice]

    return duplicado_dict

busqueda_duplicados("binario")
`````

## Ejercicio 13
Escribe una función que compruebe si un DNI es válido, devolviendo `True` en caso afirmativo.  
En primer lugar, el DNI debe tener entre 7 y 8 dígitos numéricos.  
Para saber si la letra del DNI es correcta puedes utilizar el siguiente fragmento de código:
>```Python
>DNI= 55555555
>palabra='TRWAGMYFPDXBNJZSQVHLCKE'
>print('La letra del DNI es: ', palabra[DNI%23])
>````


`````python
def comprob_dni(dni):
    letra = 'TRWAGMYFPDXBNJZSQVHLCKE'
    cant_numeros = 0
    letra_dni_dada = ''
    numeros_dni = ''
    
    for caracter in dni:
        if caracter.isnumeric():
            cant_numeros += 1
            numeros_dni = "".join([numeros_dni, caracter])
            #print(numeros_dni)
        else:
            letra_dni_dada = caracter
            #print(letra_dni_dada)
    
    letra_dni_correcta = letra[int(numeros_dni) % 23]
    #print(letra_dni_correcta)
            
    if cant_numeros == 7 or cant_numeros == 8:
        if letra_dni_dada == letra_dni_correcta:
            return True
        else:
            print("DNI incorrecto: la letra del DNI no es correcta.")
    else:
        print("DNI incorrecto: el DNI tiene un número incorrecto de caracteres numéricos.")

comprob_dni("55555555Y")
comprob_dni("55555555K")
`````
