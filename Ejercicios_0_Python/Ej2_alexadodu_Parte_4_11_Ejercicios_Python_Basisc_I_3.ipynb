## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

# Si A y B son equivalentes
print("A y B son equivalentes:", A == B)

# Si A y C NO son equivalentes
print("A y C NO son equivalentes:", A != C)

# Si A es mayor que C
print("A es mayor que C:", A > C)

# Si C es menor o igual que A
print("C es menor o igual que A:", C <= A)

# Si B NO es equivalente a C
print("B NO es equivalente a C:", B != C)
````

## Ejercicio 2
Crea un programa donde se recojan dos inputs del usuario, y el output del programa sea si esos inputs son iguales o no

````python
input1 = input("Escribe una contraseña:")
input2 = input("Escribe la contraseña nuevamente:")

if input1 == input2:
    print("¡Contraseña generada!")
else:
    print("Vaya, no coinciden. Inténtalo de nuevo.")
````

## Ejercicio 3
Mismo programa que en 2, pero en esta ocasión tienen que ser tres inputs y dos salidas. Una de las salidas que nos indique si todos son iguales, y la otra si al menos dos inputs sí que lo son

````python
input1 = input("Escribe una contraseña:")
input2 = input("Escribe la contraseña nuevamente:")
input3 = input("Esto es por MEGASEGURIDAD. Escribe la contraseña una última vez:")

if input1 == input2 == input3:
    print("Todas coinciden. ¡Contraseña generada!")
elif input1 == input2 or input1 == input3 or input2 == input3:
    print("Sólo dos de ellas coindicen. Inténtalo de nuevo.")
else:
    print("Vaya, no coincide ninguna. Inténtalo de nuevo.")
````
