## Ejercicio 2
Escribe un programa para pasar de grados a radianes. Hay que usar `input`. Recuerda que la conversión se realiza mediante

radianes = grados*(pi/180)

```python
grados_ingresados = input("Añade los grados que quieres pasar a radianes: ")

import math

radianes_calculados = int(grados_ingresados) * (math.pi / 180)
print("El resultado es:", round(radianes_calculados, 2))
```

## Ejercicio 4
Escribe un programa que calcule el area de un paralelogramo (base x altura). También con `input`. En este caso debe ser capaz de admitir valores reales, con decimales, para base y altura en español. (PISTA: Recuerda que tendrás que hacer uso del método replace)

```python
print("Para calcular el área de un paralelogramo necesito que me indiques su base y su altura.")
base = input("\tBase con decimales (cm):")
altura = input("\tAltura con decimales (cm):")

lista = [base, altura]

for indice in range(len(lista)):
    lista[indice] = lista[indice].replace(",", ".")
    lista[indice] = float(lista[indice])

area = lista[0] * lista[1]

print(
    "El área de tu paralelogramo es de",
    round(area, 2), "cm."
)
```
