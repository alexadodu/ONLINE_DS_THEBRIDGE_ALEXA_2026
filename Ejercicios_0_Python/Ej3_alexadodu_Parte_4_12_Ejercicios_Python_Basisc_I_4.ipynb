## Ejercicio 1
Para este ejercicio vamos a poner en práctica [las funciones built in](https://docs.python.org/3/library/functions.html).

1. Calcula el máximo de la lista: [4, 6, 8, -1]
2. Suma todos los elementos de la lista anterior
3. Redondea este float a 3 dígitos decimales: 63.451256965
4. Valor absoluto de: -74

```python
numeros = [4, 6, 8, -1]
# 1. Calcula el máximo de la lista: [4, 6, 8, -1]
print(max(numeros))

# 2. Suma todos los elementos de la lista anterior
print(sum(numeros))

# 3. Redondea este float a 3 dígitos decimales: 63.451256965
print(round(63.451256965, 3))

# 4. Valor absoluto de: -74
print(abs(-74))
```

## Ejercicio 2
Para el siguiente string se pide imprimir por pantalla los siguientes casos

    "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

1. Pásalo todo a mayúsculas
2. Pásalo todo a minúsculas
3. Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
4. Crea una lista dividiéndolo por sus espacios
5. Sustituye las comas `,` por puntos y comas `;`
6. Elimina las `a` minúsculas

```python
string = "En un lugar de la Mancha, de cuyo nombre no quiero acordarme."

# Pasa todo a mayúsculas
print(string.upper())

# Pasa todo a minúsculas
print(string.lower())

# Solo la primera letra de cada palabra en mayúscula, el resto en minúscula
print(string.title())

# Crea una lista dividiéndolo por sus espacios
lista = string.split(" ")
print(lista)

# Sustituye las comas , por puntos y comas ;
print(string.replace(",", ";"))

# Elimina las a minúsculas
print(string.replace("a", ""))
```

## Ejercicio 3
1. Crea una lista con 3 elementos numéricos
2. Añade un cuarto elemento
3. Calcula la suma de todos
4. Elimina el segundo elemento de la lista
5. Añade otro elemento en la posicion 3 de la lista
6. Crea otra lista con 4 elementos y concaténala a la que ya tenías.
7. Ordena la lista de menor a mayor.
8. BONUS: Ordena la lista de mayor a menor.  
(*PISTA: Para hacer 8, tendrás que buscar en la información del método...*)  
  
  
NOTA: Pinta por pantalla el resultado de cada paso.

```python
lista = [1, 2, 3]
print("Lista inicial:", lista)

lista.append(4)
print("Lista actualizada:", lista)

print("Suma de todos los elementos de la lista:", sum(lista))

lista.pop(1)
print("Lista actualizada, eliminando el segundo elemento:", lista)

lista.insert(2, 200)
print("Lista actualizada, añadiendo el valor de 200 en la 3a posición:", lista)

lista_2 = [100, 101, 102, 103]
lista_mix = lista + lista_2
print("Lista mix como resultado de la concatenación de la primera y una nueva generada de 4 elementos:", lista_mix)

lista_mix.sort()
print("Lista mix ordenada de menor a mayor:", lista_mix)

lista_mix.sort(reverse = True)
print("Lista mix ordenada de mayor a menor:", lista_mix)
```
